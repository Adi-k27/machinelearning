```python
#Import FastAPI
from fastapi import FastAPI

app = FastAPI()

@app.get("/rule/{age}/{gender}")

def royal_identity(age: int, gender: str):
    if gender.lower() == "female":
        if age < 20:
            return {"output":"You are a Princess"}
        else:
            return {"output":"You are a Queen"}
    else:
        if age < 20:
            return {"output":"You are a Prince"}
        else:
            return {"output":"You are a King"}
```

#### We need a server to run this application. `Uvicorn` is a recomended server for running fastapi applications

```python
# lets us start the uvicorn server
# before that if we run uvicorn sever with jupyter notebook it may make the notebook unresponsive.
# therefore we will use nest_asyncio to ensure uvicorn runs in a different process/thread
import nest_asyncio
from uvicorn import run

# this makes sure the uvicorn sever runs in a seperate process and not interfere with Jupyter notebook process.
nest_asyncio.apply()
run(royal_identity, host="localhost", port=8000)
```
